// C++ code for Automatic Night Lamp

int led1 = 13;

void setup() {
  Serial.begin(9600);
  pinMode(led1, OUTPUT);
}

void loop() {
  int c = analogRead(A0);   // <-- corrected this line
  Serial.println(c);

  if (c < 800) {
    digitalWrite(led1, LOW);   // Light OFF (day time)
  } else {
    digitalWrite(led1, HIGH);  // Light ON (night time)
  }
}
