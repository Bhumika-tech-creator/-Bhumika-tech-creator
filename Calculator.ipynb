```python
import ipywidgets as widgets
from IPython.display import display, clear_output

output = widgets.Output()
screen = widgets.Text(value="", disabled=False, layout=widgets.Layout(width='400px'))

def on_button_click(b):
    with output:
        clear_output()
        text = b.description
        if text == "=":
            try:
                screen.value = str(eval(screen.value))
            except:
                screen.value = "Error"
        elif text == "C":
            screen.value = ""
        else:
            screen.value += text
        print(screen.value)

buttons = [
    ["7", "8", "9", "/"],
    ["4", "5", "6", "*"],
    ["1", "2", "3", "-"],
    ["C", "0", "=", "+"]
]

button_widgets = []
for row in buttons:
    row_buttons = []
    for label in row:
        btn = widgets.Button(description=label, layout=widgets.Layout(width='90px'))
        btn.on_click(on_button_click)
        row_buttons.append(btn)
    button_widgets.append(widgets.HBox(row_buttons))

display(screen)
for row in button_widgets:
    display(row)
display(output)
```
